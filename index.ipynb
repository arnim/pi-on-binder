# Quick tour: pi-on-binder 

Welcome! This Binder ships with **[pi-coding-agent](https://www.npmjs.com/package/@earendil-works/pi-coding-agent)**, **Python 3.11**, and **R 4.4** - everything you need for AI-assisted coding, data exploration and modelling.

---

## 🤖 pi-coding-agent

An instance of pi should have automatically started in a panel.

### From Jupyter launcher

To start a new instance of pi, click the **Pi Agent** icon (with the pi logo) in the Jupyter launcher.
It opens a terminal and runs pi with `pi --model openrouter/deepseek/deepseek-v4-pro --approve` automatically.

Alternatively, open a terminal and type `pi`.

## 🍕 Try out the pi

Enter the following prompts in `pi`:

1. Do a (simple) multiverse analysis of kaggle titanic dataset and create a specification curve.

2. Create a short beginner-friendly Jupyter notebook tutorial in R (preferably with httr2/jsonlite) that teaches how to call LLM APIs through OpenRouter. Include a minimal "Hello world" demo using the environment variable OPENROUTER_API_KEY and meta-llama/llama-3.2-1b-instruct. Test the notebook to verify that Jupyter notwebook works.
