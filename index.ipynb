## 🔬 Python quick start

Create a new notebook, select the **Python 3** kernel, and go:

````python
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns

df = pd.DataFrame({'x': np.linspace(0, 10, 50)})
df['y'] = np.sin(df['x']) + np.random.randn(50) * 0.1

sns.lineplot(data=df, x='x', y='y')
plt.title('Hello from Python!')
plt.show()
````

---

## 📊 R quick start

Create a new notebook, select the **R** kernel:

````python
library(ggplot2)

df <- data.frame(x = seq(0, 10, length.out = 50))
df$y <- sin(df$x) + rnorm(50, sd = 0.1)

ggplot(df, aes(x, y)) +
  geom_line(colour = 'steelblue', size = 1) +
  ggtitle('Hello from R!')
````

## 🤖 pi-coding-agent

### From a terminal

Open a **terminal** in JupyterLab (`File → New → Terminal`) and run:

```bash
export ANTHROPIC_API_KEY=sk-ant-...   # your API key
pi                                    # interactive mode
pi -p "Analyze the file data.csv"     # one-shot
```

### From a notebook cell

Create a new notebook and select the **"Pi Agent"** kernel. Each cell you run
is sent to pi as a prompt.

**Kernel magics:**

| Magic | Action |
|-------|--------|
| `%pi_models` | List available models |
| `%pi_model <id>` | Switch model (e.g. `anthropic/claude-sonnet-4-5`) |
| `%pi_thinking high` | Set thinking level |
| `%pi_new_session` | Start a fresh session |
| `%pi_compact` | Compact long context |
| `%pi_help` | Show all magics |
